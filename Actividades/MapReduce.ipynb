```python
from collections import defaultdict

# INPUT
lineas = [
    "Hola mundo",
    "Hola Oscar",
    "Mundo feliz"
]
```

```python
# -------------------------
# MAP
# -------------------------
pares = []
for linea in lineas:
    for palabra in linea.lower().split():
        pares.append((palabra, 1))
```

```python
# -------------------------
# SHUFFLE & SORT
# Redistribución y ordenamiento
# -------------------------
grupos = defaultdict(list)
for palabra, valor in pares:
    grupos[palabra].append(valor)
```

```python
# -------------------------
# REDUCE
# -------------------------
resultado = {}
for palabra in sorted(grupos):
    resultado[palabra] = sum(grupos[palabra])
```

```python
# OUTPUT
print("Conteo final de palabras:")
for palabra, cantidad in resultado.items():
    print(palabra, ":", cantidad)
```
